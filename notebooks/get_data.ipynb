```python
import os
import json

from sec_edgar_downloader import Downloader
```

```python
with open('../cfg.json', 'r') as f:
    config = json.load(f)
ticker_list = config["ticker_list"]
company_name = config["company_name"]
email = config["email"]
write_folder_path = config['initial_write_path']
```

```python
# create the destination folder if it doesn't exist
os.makedirs(write_folder_path, exist_ok=True)
```

```python
dl = Downloader(company_name, email, write_folder_path)

for ticker in ticker_list:
    dl.get("10-K", ticker, after="2024-01-01", download_details=False)
```
